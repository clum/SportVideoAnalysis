# Trim a proxy to selected ranges

This notebook joins the time ranges entered in `Test/segments_template.xlsx` into one shorter clip from the Step 1 proxy. Fill at least one row on the workbook's `Segments` tab before running the trimming cell. The workbook uses elapsed seconds from the beginning of the proxy, with start-inclusive and stop-exclusive ranges.

Install the Python requirements with `python -m pip install -r Utilities/requirements.txt`. FFmpeg and ffprobe must also be installed and available on `PATH`. The output is written to `output/clips/`, which is ignored by Git.

```python
from pathlib import Path
import sys

starting_directory = Path.cwd().resolve()
repo_root = next(
    (directory for directory in (starting_directory, *starting_directory.parents)
     if (directory / "Utilities" / "trim_proxy_video.py").is_file()),
    None,
)
if repo_root is None:
    raise RuntimeError("Could not find the SportVideoAnalysis repository root above the current directory.")
sys.path.insert(0, str(repo_root / "Utilities"))

from trim_proxy_video import trim_proxy
```

```python
proxy_video = repo_root / "output" / "proxies" / "26_10_24_GoPro_Goal_proxy.mp4"
ranges_workbook = repo_root / "Test" / "segments_template.xlsx"
trimmed_video = repo_root / "output" / "clips" / "26_10_24_GoPro_Goal_selected.mp4"

print(f"Proxy: {proxy_video}")
print(f"Ranges workbook (fill the Segments tab first): {ranges_workbook}")
print(f"Output: {trimmed_video}")
```

```python
result = trim_proxy(
    proxy_video,
    ranges_workbook,
    trimmed_video,
    include_audio=True,
    overwrite=True,
)

result
```
