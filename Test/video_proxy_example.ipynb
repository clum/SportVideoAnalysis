# Create a video proxy

This example calls the reusable `create_proxy` function with the soccer sample in this repository. Run Jupyter from the repository root. Install the Python notebook dependency with `python -m pip install -r Utilities/requirements.txt`. FFmpeg and ffprobe must also be installed and available on `PATH`.

The output is written under `output/proxies/`, a generated folder ignored by Git. The call opts into overwriting that output so the notebook can be rerun.

```python
from pathlib import Path
import sys

repo_root = Path.cwd().resolve()
if not (repo_root / "AGENTS.md").is_file():
    raise RuntimeError("Start Jupyter from the SportVideoAnalysis repository root.")
sys.path.insert(0, str(repo_root / "Utilities"))

from video_proxy import create_proxy
```

```python
source_video = repo_root / "Test" / "Soccer" / "26_10_24_GoPro_Goal.mp4"
proxy_video = repo_root / "output" / "proxies" / "26_10_24_GoPro_Goal_proxy.mp4"

result = create_proxy(
    source_video,
    proxy_video,
    max_width=1280,
    max_height=720,
    target_fps=15,
    include_audio=True,
    overwrite=True,
)

print(result)
```
