# Find additional key-moment candidates

This example scans the Step 2 trimmed soccer clip using the `soccer` profile and the `KeyMoments` annotation workbook. Known labels are used only as annotations; newly found candidates are returned without event labels. The current detector uses OpenCV motion and frame-change cues, so its score is a heuristic and should be reviewed.

Install packages with `python -m pip install -r Utilities/requirements.txt`. FFmpeg/ffprobe must also be installed and on `PATH`. No model weights are required for this baseline.

```python
from pathlib import Path
import sys

starting_directory = Path.cwd().resolve()
repo_root = next(
    (directory for directory in (starting_directory, *starting_directory.parents)
     if (directory / 'Utilities' / 'detect_key_moments.py').is_file()),
    None,
)
if repo_root is None:
    raise RuntimeError('Could not find the SportVideoAnalysis repository root above the current directory.')
sys.path.insert(0, str(repo_root / 'Utilities'))

from detect_key_moments import SportType, detect_key_moments
```

```python
# The workbook's example times/labels are illustrative. Review them against the clip first.
trimmed_video = repo_root / 'output' / 'clips' / '26_10_24_GoPro_Goal_selected.mp4'
sport = SportType.SOCCER
examples_workbook = repo_root / 'Test' / 'key_moments_test1.xlsx'
results_workbook = repo_root / 'output' / 'clips' / '26_10_24_GoPro_Goal_selected_key_moments.xlsx'

print(f'Video: {trimmed_video}')
print(f'Sport: {sport.value}')
print(f'Known examples: {examples_workbook}')
print(f'Results: {results_workbook}')
```

```python
result = detect_key_moments(
    trimmed_video,
    sport,
    examples_workbook,
    results_workbook,
    overwrite=True,  # set False to refuse replacing an existing results workbook
)

print(f'Found {len(result.detections)} additional candidates from {result.example_count} examples.')
print(f'Results workbook: {result.output_path}')
result.detections
```
